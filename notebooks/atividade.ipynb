Quatro curvas de aprendizado, CSVs auditáveis e PDF. Execute as células em ordem. Para o experimento completo, use GPU. A configuração rápida é exploratória; o PDF explicita seu escopo.


```python
from pathlib import Path
import os, subprocess, sys
if not Path('pyproject.toml').exists():
    if Path('../pyproject.toml').exists():
        os.chdir('..')
    else:
        subprocess.run(['git', 'clone', 'https://github.com/tpenha05/nlp-finetuning-distilbert.git'], check=True)
        os.chdir('nlp-finetuning-distilbert')
subprocess.run([sys.executable, '-m', 'pip', 'install', '-e', '.'], check=True)
```

```python
import torch
print('GPU disponível:', torch.cuda.is_available())
PROFILE = 'quick'  # altere para 'full' para usar o teste oficial completo
DATASET = 'imdb'  # bônus: 'ag_news'
from datetime import datetime
out = Path('results/custom') / f'{DATASET}_{PROFILE}_{datetime.now():%Y%m%d_%H%M%S}'
subprocess.run([sys.executable, '-m', 'bert_curves.experiment', '--config',
                f'configs/{PROFILE}.json', '--dataset', DATASET, '--output', str(out)], check=True)
```

```python
from IPython.display import display, Image, FileLink
import pandas as pd
display(Image(filename=str(out / 'learning_curves.png')))
display(pd.read_csv(out / 'summary.csv'))
display(FileLink(str(out / 'relatorio.pdf')))
```

## Como interpretar

Compare o crescimento e a variabilidade das quatro curvas. O modelo do zero tem o mesmo encoder, mas não os pesos pré-treinados. Congelar treina apenas a cabeça; fine-tuning adapta tudo. Uma curva não precisa ser monotônica. Use as medidas observadas para responder à atividade, respeitando o tamanho do teste e o orçamento de otimização.
